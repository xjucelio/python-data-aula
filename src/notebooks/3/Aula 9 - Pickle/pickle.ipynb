# Tratamento dos Dados

## Quebrando a Variável de Data

```python
# 1. Converter a coluna de datas para o formato datetime do pandas
# O argumento 'format' garante que ele leia 'DD/MM/AAAA' corretamente.
dados['Date_of_Journey'] = pd.to_datetime(dados['Date_of_Journey'], format='%d/%m/%Y')

# 2. Extrair dia, mês e ano usando o acessor .dt (muito mais rápido!)
dados['day'] = dados['Date_of_Journey'].dt.day
dados['month'] = dados['Date_of_Journey'].dt.month
dados['year'] = dados['Date_of_Journey'].dt.year

# Agora as suas colunas 'day', 'month' e 'year' estão preenchidas corretamente.
print(dados.head())
```

```python
dados
```

```python
dados = dados.drop('Date_of_Journey', axis = 1)
```

## Quebrando a variável de Hora (Partida e Chegada)

```python
dados["departure_hour"] = dados["Dep_Time"].str[0:2].astype(int)
dados["departure_min"]  = dados["Dep_Time"].str[3:5].astype(int)

# Cria as colunas a partir de Arrival_Time
dados["arrival_hour"] = dados["Arrival_Time"].str[0:2].astype(int)
dados["arrival_min"]  = dados["Arrival_Time"].str[3:5].astype(int)
```

```python
dados
```

```python
dados = dados.drop('Dep_Time', axis = 1)
dados = dados.drop('Arrival_Time', axis = 1)
```

```python
dados
```

Agora que temos algumas variáveis novas na nossa base de dados, vamos tratar as variáveis categóricas

## Tratando a variável de quantidade de paradas

```python
dados.Total_Stops.unique()
```

```python
dados['stops'] = dados['Total_Stops'].str.extract(r'(\d+)').fillna(0).astype(int)

# Verificar o resultado
print(dados[['Total_Stops', 'stops']].head())
```

```python
dados
```

```python
dados = dados.drop('Total_Stops', axis = 1)
```

# Aplicando o Label Encoder

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

colunas = dados.dtypes.reset_index()

categ_cols = colunas[colunas[0] == 'object']['index'].to_list()


# Preenchendo as variáveis com dados nulos com a média


for i in categ_cols:
    dados[str(i) +'_encoded'] = le.fit_transform(dados[i])
    dados = dados.drop(i,axis = 1)
```

```python
dados
```

# Quebrando o dataset para modelagem

```python
from sklearn.model_selection import train_test_split

x = dados.drop('Price', axis = 1)
y = dados['Price']

x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.3, random_state=42)
```

# Carregando o modelo

Iremos fazer o mesmo processo para os 3 modelos que falamos para verificar as diferenças

```python
from sklearn.ensemble import GradientBoostingRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from sklearn.metrics import r2_score

import time

modelo_gr = GradientBoostingRegressor(
    max_depth=10,
    max_leaf_nodes=20,
    random_state=42
)

modelo_xb = XGBRegressor(
    max_depth=10,
    grow_policy="lossguide",  # Para usar max_leaves
    max_leaves=20,
    random_state=42
)

modelo_lg = LGBMRegressor(
    num_leaves=31,       # mais folhas, mais complexidade do modelo
    max_depth=-1,        # deixa ilimitado, só num_leaves controla
    learning_rate=0.05,  # menor taxa, melhora generalização
    n_estimators=1000,   # mais árvores
    boosting_type='gbdt',
    random_state=42
)



print('=========== Gradient Boosting ==============')

start = time.time()
modelo_gr.fit(x_train, y_train)
end = time.time()
y_pred_train = modelo_gr.predict(x_train)
y_pred_test = modelo_gr.predict(x_test)

print(f'O resultado na base de treino é: {r2_score(y_train,y_pred_train):.3f}')
print(f'O resultado na base de teste é: {r2_score(y_test,y_pred_test):.3f}')
print(f'O tempo que o modelo demorou para treinar foi: {end-start:.2f}')

print()
print('=========== XG Boost ==============')

start = time.time()
modelo_xb.fit(x_train, y_train)
end = time.time()
y_pred_train = modelo_xb.predict(x_train)
y_pred_test = modelo_xb.predict(x_test)


print(f'O resultado na base de treino é: {r2_score(y_train,y_pred_train):.3f}')
print(f'O resultado na base de teste é: {r2_score(y_test,y_pred_test):.3f}')
print(f'O tempo que o modelo demorou para treinar foi: {end-start:.2f}')

print()
print('=========== LGBM ==============')

start = time.time()
modelo_lg.fit(x_train, y_train)
end = time.time()
y_pred_train = modelo_lg.predict(x_train)
y_pred_test = modelo_lg.predict(x_test)

print(f'O resultado na base de treino é: {r2_score(y_train,y_pred_train):.3f}') # type: ignore
print(f'O resultado na base de teste é: {r2_score(y_test,y_pred_test):.3f}') # type: ignore
print(f'O tempo que o modelo demorou para treinar foi: {end-start:.2f}')

```

É possível ver as principais diferenças entre os modelos

- O GradientBoosting é um bom parâmetro para início

- O XGBoost traz resultados mais precisos

- O LGBM é treinado com uma velocidade muito maior

Mas o que significam wb e rb?


Basicamente é forma que estamos usando o arquivo.


Quando estamos usando o método dump, estamos escrevendo(salvando) um arquivo, então wb significa write byte. Em resumo estamos passando privilégios de escrita.
No caso do carregamento estamos fazendo leitura, então usando read ao invés de write.


No momento de dizer o nome do arquivo como "modelo.sav" podemos passar uma path ou caminho inteiro no computador, para salvar o modelo em alguma pasta específica. Neste caso passamos apenas o nome do arquivo, isso significa que o modelo será salvo na mesma pasta que o notebook.


O modelo sempre é salvo pós treinamento, então ele já guarda tudo que aprendeu com o treinamento.

Use sempre esta alternativa quando conseguir um modelo com bons resultados para não ter que passar pelo tempo de treinamento novamente, principalmente quando for necessário um grande tempo para treinar seu modelo.

Dessa forma é possível automatizar fluxos de dados, chamando direto o modelo treinado e pedindo um resultado predict, para saber a predição. Sem precisar treinar o modelo com novos dados toda vez.

Mesmo assim é uma boa prática de tempos em tempos rever se seu modelo ainda está performando bem e se não precisa ser treinado novamente, principalmente quando a característica dos seus dados vai mudando ao longo do tempo.
