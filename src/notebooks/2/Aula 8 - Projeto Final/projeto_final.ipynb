## Verificando outliers nas variáveis contínuas

```python
def detectar_outliers_iqr(df, colunas):
    """Detecta outliers usando método IQR de forma vetorizada"""
    resultados = []
    
    for col in colunas:
        q1, q3 = df[col].quantile([0.25, 0.75])
        iqr = q3 - q1
        li, ls = q1 - 1.5 * iqr, q3 + 1.5 * iqr
        
        mask_outliers = (df[col] < li) | (df[col] > ls)
        perc_outliers = (mask_outliers.sum() / len(df)) * 100
        
        resultados.append({
            'coluna': col,
            'perc_outliers': perc_outliers,
            'qtt_outliers': mask_outliers.sum()
        })
    
    return pd.DataFrame(resultados)

# Uso
df_outliers = detectar_outliers_iqr(df, cont_cols.coluna)
df_outliers[df_outliers.perc_outliers > 10]
```

## Separando o Dataset entre Treino e Teste

```python
from sklearn.model_selection import train_test_split

x = df.drop('round_winner_encoded', axis = 1)
y = df.round_winner_encoded

```

### Padronizando as variáveis de entrada do modelo

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_scaled = scaler.fit_transform(x)
```

```python
x_scaled
```

```python
x_scaled = pd.DataFrame(x_scaled, columns=(x.columns))
```

```python
x_scaled
```

### Quebrando o Dataframe

```python
x_train, x_test, y_train, y_test = train_test_split(x_scaled,y , test_size=0.2, random_state=42)
```

## Importando o modelo

```python
from sklearn.linear_model import LogisticRegression

modelo = LogisticRegression()

modelo.fit(x_train, y_train)



y_pred = modelo.predict(x_test)
```

```python
y_pred
```

## Avaliando o Modelo

```python
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix


print('acurácia do modelo foi : ' + str(accuracy_score(y_test, y_pred)))
print('f1 score do modelo foi : ' + str(f1_score(y_test, y_pred)))
cm = confusion_matrix(y_test, y_pred)
```
